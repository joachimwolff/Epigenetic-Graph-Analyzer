```python
import os

folder_path = ROOT_PATH  # Replace with the actual folder path

files = [os.path.join(ROOT_PATH, f) for f in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, f))]

```

```python
def restoreRestrictionCutSites(list_of_pd_interactions_path):

    stacked_df = None
    for i, pd_interactions_path in enumerate(list_of_pd_interactions_path):
        pd_interactions = pd.read_table(pd_interactions_path, header=None, sep='\t|,', compression='gzip')
        data_frame = pd_interactions[[0,1,2,3,4,5]]       
        data_frame_forward = data_frame[[0,1,2]].rename(columns={0: 'Chromosome', 1: 'Start', 2: 'End'})
        data_frame_reverse = data_frame[[3,4,5]].rename(columns={3: 'Chromosome', 4: 'Start', 5: 'End'})
        if stacked_df is None:
            stacked_df = pd.concat([data_frame_forward, data_frame_reverse], axis=0, ignore_index=True)
        else:
            stacked_df = pd.concat([stacked_df, data_frame_forward, data_frame_reverse], axis=0, ignore_index=True)


    stacked_df.drop_duplicates(inplace=True, ignore_index=True)
    stacked_df.sort_values(by=['Chromosome', 'Start', 'End'], inplace=True)
    stacked_df.reset_index(inplace=True, drop=True)
    return stacked_df
```

```python
restriction_cut_sited_df = restoreRestrictionCutSites(files)
```

```python
restriction_cut_sited_df.to_csv(os.path.join(ROOT_PATH + 'restriction_cut_sites.csv'), index=False, sep='\t')
```
