```python
def check_md5(args):
    file_name, folder_path = args
    if file_name.endswith('.md5'):
        return

    md5_file = file_name + '.md5'
    md5_path = os.path.join(folder_path, md5_file)
    # print(md5_path + '\n')

    if not os.path.isfile(md5_path):
        print(f"MD5 file not found for {file_name}\n")
        return

    with open(md5_path, 'r') as f:
        expected_md5 = f.read().strip().split()[0]

    with open(os.path.join(folder_path, file_name), 'rb') as f:
        file_data = f.read()
        calculated_md5 = hashlib.md5(file_data).hexdigest()

    if expected_md5 != calculated_md5:
        print(f"{file_name}: MD5 sum does not match")
        print(f"{file_name}: {expected_md5}")
        print(f"{file_name}: {calculated_md5}")
```
