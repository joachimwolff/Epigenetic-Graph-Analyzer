```python
# df = pd.ExcelFile(f'{ROOT_PATH}/41588_2015_BFng3286_MOESM48_ESM.xlsx').parse('Suppl Table 4')
df = pd.read_excel(os.path.join(ROOT_PATH, '41588_2015_BFng3286_MOESM48_ESM.xlsx'), sheet_name='Suppl Table 4', header=None)
```

```python
df = df[df[0].str.startswith('>')]
```

```python
df[["position", "strand", "Name"]] = df[0].str.split(' ', expand=True)
```

```python
df[["Chromosome", "Start", "End"]] = df["position"].str.split(":|-", n=2, expand=True)
```

```python
df['Chromosome'] =  df['Chromosome'].str.replace('>', '').apply(lowercase_first_letter)
```

```python
df[["Chromosome", 'Start', "End", "Name"]].to_csv(os.path.join(ROOT_PATH, 'promoters_mifsud.tab'), index=False, sep='\t')
```

```python
def lowercase_first_letter(s):
    return s[:1].lower() + s[1:]
```
